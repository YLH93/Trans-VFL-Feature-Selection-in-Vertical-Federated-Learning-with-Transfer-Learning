# Trans-VFL EHR Simulation

Synthetic high-dimensional EHR setting from Section 4.1 of the PDF. The default demo uses a smaller patient count than the full paper setting so it runs quickly.

This notebook uses the reusable NumPy implementation in `trans_vfl`. It mirrors the PDF workflow: frozen learned features, server-side embedding component selection, and local Stage 3 group-lasso pruning.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
```

```python
from trans_vfl.simulations import EHR_PAPER_TARGET, EHRSimulationConfig, run_ehr_simulation
```

```python
config = EHRSimulationConfig(
    n_sites=20,
    n_patients=5000,
    replications=25,
)
result = run_ehr_simulation(config, seed=0)
result
```

```python
print('metric                  observed    pdf_target')
print(f"accuracy                {result['accuracy']:.4f}      {EHR_PAPER_TARGET['accuracy']:.4f}")
print(f"fdr                     {result['fdr']:.4f}      {EHR_PAPER_TARGET['fdr']:.4f}")
print(f"tpr                     {result['tpr']:.4f}      {EHR_PAPER_TARGET['tpr']:.4f}")
print(f"tnr                     {result['tnr']:.4f}      {EHR_PAPER_TARGET['tnr']:.4f}")
print(f"avg_features_selected   {result['avg_features_selected']:.2f}       {EHR_PAPER_TARGET['avg_features_selected']:.2f}")
```
